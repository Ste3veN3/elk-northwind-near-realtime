```python
from pymongo import MongoClient
import random
import time
from datetime import datetime, timedelta

# Conexión con MongoDB (sin autenticación, igual que en tu docker-compose)
client = MongoClient("mongodb://localhost:27017/")
db = client["northwind_mongo"]
collection = db["employees"]

# Calcula el siguiente employeeID (no es autoincremental en Mongo)
ultimo = collection.find_one(sort=[("employeeID", -1)])
siguiente_id = (ultimo["employeeID"] if ultimo else 0) + 1

titulos = [
    "Sales Representative", "Sales Manager",
    "Vice President, Sales", "Inside Sales Coordinator"
]
nombres = ["Carlos", "Maria", "Luis", "Ana", "Jorge", "Paula", "Diego", "Sofia"]
apellidos = ["Perez", "Gomez", "Diaz", "Torres", "Ramirez", "Castro"]
ciudades = [
    ("Seattle", "WA", "98122", "USA"),
    ("London", None, "SW1P 2AR", "UK"),
    ("Quito", "Pichincha", "170101", "Ecuador"),
]

print("======================================")
print(" Generador de empleados (Mongo)")
print("======================================")
print("Insertando un empleado nuevo cada 15 segundos")
print("Usa el botón de stop del kernel para detenerlo.\n")

try:
    while True:
        first = random.choice(nombres)
        last = random.choice(apellidos)
        ciudad, region, cp, pais = random.choice(ciudades)
        hire_date = datetime.now() - timedelta(days=random.randint(0, 10))
        birth_date = datetime(1970, 1, 1) + timedelta(days=random.randint(0, 15000))

        doc = {
            "employeeID": siguiente_id,
            "lastName": last,
            "firstName": first,
            "title": random.choice(titulos),
            "titleOfCourtesy": random.choice(["Mr.", "Ms.", "Mrs."]),
            "birthDate": birth_date.strftime("%Y-%m-%d %H:%M:%S.000"),
            "hireDate": hire_date.strftime("%Y-%m-%d %H:%M:%S.000"),
            "address": "Calle Simulada 123",
            "city": ciudad,
            "region": region,
            "postalCode": cp,
            "country": pais,
            "homePhone": "(555) 555-0000",
            "extension": str(random.randint(1000, 9999)),
            "notes": "Empleado generado automaticamente para la demo.",
            "reportsTo": 2,
            "photoPath": "http://accweb/emmployees/default.bmp",
        }

        collection.insert_one(doc)

        print(
            f"[{datetime.now().strftime('%H:%M:%S')}] "
            f"employeeID: {siguiente_id} | {first} {last} | {doc['title']}"
        )

        siguiente_id += 1
        time.sleep(15)

except KeyboardInterrupt:
    print("\nProceso detenido.")

finally:
    client.close()
    print("Conexión Mongo cerrada.")
```
